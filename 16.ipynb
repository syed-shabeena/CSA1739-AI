```python
import numpy as np

x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
w = np.random.rand(2, 1)
b = 0.1

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

output = sigmoid(np.dot(x, w) + b)

print("Network Output:")
print(output)
```
