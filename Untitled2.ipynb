```python
import random

# Grid environment
rows = 4
cols = 4

start = (0, 0)
goal = (3, 3)

obstacles = [(1, 1), (2, 1), (2, 2)]

actions = {
    0: (-1, 0),  # Up
    1: (1, 0),   # Down
    2: (0, -1),  # Left
    3: (0, 1)    # Right
}

# Initialize Q-table
Q = {}

alpha = 0.1
gamma = 0.9
epsilon = 0.2
episodes = 5000

def valid_actions(state):
    valid = []

    for action, (dr, dc) in actions.items():
        nr = state[0] + dr
        nc = state[1] + dc

        if (0 <= nr < rows and
            0 <= nc < cols and
            (nr, nc) not in obstacles):
            valid.append(action)

    return valid

def next_state(state, action):
    dr, dc = actions[action]
    return (state[0] + dr, state[1] + dc)

# Q-Learning training
for episode in range(episodes):

    state = start

    for step in range(100):

        valid = valid_actions(state)

        if random.random() < epsilon:
            action = random.choice(valid)
        else:
            action = max(
                valid,
                key=lambda a: Q.get((state, a), 0)
            )

        new_state = next_state(state, action)

        if new_state == goal:
            reward = 100
        else:
            reward = -1

        old_value = Q.get((state, action), 0)

        next_valid = valid_actions(new_state)

        if new_state == goal:
            future = 0
        else:
            future = max(
                Q.get((new_state, a), 0)
                for a in next_valid
            )

        Q[(state, action)] = old_value + alpha * (
            reward + gamma * future - old_value
        )

        state = new_state

        if state == goal:
            break

# Extract learned path
state = start
path = [state]
visited = {state}

for step in range(20):

    if state == goal:
        break

    valid = valid_actions(state)

    action = max(
        valid,
        key=lambda a: Q.get((state, a), 0)
    )

    state = next_state(state, action)

    if state in visited:
        break

    path.append(state)
    visited.add(state)

print("Q-LEARNING OPTIMAL PATH")
print("Start:", start)
print("Goal:", goal)
print("Obstacles:", obstacles)
print("Learned Path:", path)

if state == goal:
    print("Goal Reached Successfully!")
else:
    print("Path not found. Train again.")
```

```python
import math

board = [' '] * 9

def display():
    for i in range(0, 9, 3):
        print('|'.join(board[i:i+3]))
        if i < 6:
            print("-----")

def winner(b):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, c, d in wins:
        if b[a] == b[c] == b[d] and b[a] != ' ':
            return b[a]
    return None

def minimax(b, depth, alpha, beta, maximizing):
    w = winner(b)

    if w == 'O':
        return 10 - depth
    if w == 'X':
        return depth - 10
    if ' ' not in b:
        return 0

    if maximizing:
        best = -math.inf

        for i in range(9):
            if b[i] == ' ':
                b[i] = 'O'
                score = minimax(b, depth+1, alpha, beta, False)
                b[i] = ' '

                best = max(best, score)
                alpha = max(alpha, best)

                if beta <= alpha:
                    break

        return best

    else:
        best = math.inf

        for i in range(9):
            if b[i] == ' ':
                b[i] = 'X'
                score = minimax(b, depth+1, alpha, beta, True)
                b[i] = ' '

                best = min(best, score)
                beta = min(beta, best)

                if beta <= alpha:
                    break

        return best

def computer_move():
    best_score = -math.inf
    move = -1

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            score = minimax(board, 0, -math.inf,
                            math.inf, False)
            board[i] = ' '

            if score > best_score:
                best_score = score
                move = i

    if move != -1:
        board[move] = 'O'

print("TIC TAC TOE")
print("You are X, Computer is O")
print("Positions are numbered 1 to 9")

while True:
    display()

    try:
        pos = int(input("Enter position (1-9): ")) - 1
    except ValueError:
        print("Enter a valid number.")
        continue

    if pos < 0 or pos >= 9 or board[pos] != ' ':
        print("Invalid position! Try again.")
        continue

    board[pos] = 'X'

    if winner(board):
        display()
        print("You Win!")
        break

    if ' ' not in board:
        display()
        print("Game Draw!")
        break

    computer_move()

    if winner(board):
        display()
        print("Computer Wins!")
        break

    if ' ' not in board:
        display()
        print("Game Draw!")
        break
```

```python
from collections import deque
import heapq

graph = {
    'S': [('A', 1), ('B', 4)],
    'A': [('C', 2), ('D', 5)],
    'B': [('D', 1)],
    'C': [('G', 5)],
    'D': [('G', 3)],
    'G': []
}

heuristic = {
    'S': 7, 'A': 6, 'B': 4,
    'C': 4, 'D': 3, 'G': 0
}

def bfs(start, goal):
    queue = deque([(start, [start])])
    visited = set()

    while queue:
        node, path = queue.popleft()

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor, cost in graph[node]:
                if neighbor not in visited:
                    queue.append((neighbor, path + [neighbor]))

    return None

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        node, path = stack.pop()

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor, cost in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))

    return None

def astar(start, goal):
    pq = [(heuristic[start], 0, start, [start])]
    visited = set()

    while pq:
        f, g, node, path = heapq.heappop(pq)

        if node == goal:
            return path, g

        if node in visited:
            continue

        visited.add(node)

        for neighbor, cost in graph[node]:
            if neighbor not in visited:
                new_g = g + cost
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(
                    pq, (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, float('inf')

start = 'S'
goal = 'G'

print("BFS Path:", bfs(start, goal))
print("DFS Path:", dfs(start, goal))

path, cost = astar(start, goal)
print("A* Path:", path)
print("A* Cost:", cost)
```
